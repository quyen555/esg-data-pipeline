```python
spark.sql("CREATE SCHEMA IF NOT EXISTS workspace.bronze")
```

```python
import pandas as pd

SOURCE_FILE = "../data/raw/utility_kpi_raw.csv"

pdf = pd.read_csv(SOURCE_FILE, dtype=str, keep_default_na=False)
print(pdf.shape)
pdf.head()
```

```python
from pyspark.sql import functions as F

df = (
    spark.createDataFrame(pdf)
    .withColumn("_ingested_at", F.current_timestamp())
    .withColumn("_source_file", F.lit("utility_kpi_raw.csv"))
)

df.write.mode("overwrite").saveAsTable("workspace.bronze.utility_kpis")
```

```python
bronze = spark.table("workspace.bronze.utility_kpis")
print(bronze.count())
display(bronze)
```
